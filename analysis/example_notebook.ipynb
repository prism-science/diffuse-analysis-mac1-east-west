```python
from pathlib import Path
import yaml

from mdx2.io import loadobj
import matplotlib.pyplot as plt

# load the site config
with open("site_config.yaml") as f:
    config = yaml.safe_load(f)

# project directory
project_directory = Path(config['project_directory'])

# load some data from a file
scaling_model = loadobj(project_directory / "chess/mac1_4/mdx2/scales.nxs", 'scaling_model')

# plot the scale factors vs. phi
scaling_model.to_xarray().plot()
```
