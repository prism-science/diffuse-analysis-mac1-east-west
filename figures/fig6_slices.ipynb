```python
import xarray as xr
import json

# load source data

DATASETS = ['CHESS 2025-11-05','ALS 2026-04-23']

fn = 'source_data/fig6_slices.json'
with open(fn,'r') as f:
    print('loading',fn)
    data = xr.Dataset.from_dict(json.load(f))
```

```python
import matplotlib.pyplot as plt

dataset_legends = {
    'CHESS 2025-11-05': 'CHESS 2025-11-05\n(large xtals)',
    'ALS 2026-04-23': 'ALS 2026-04-23\n(small xtals)',
}

plot_opts = dict(
    x="h", 
    y="k", 
    vmin=-0.12, 
    vmax=0.12, 
    cmap='RdBu_r',
    add_colorbar=False,
    rasterized=True,
)
fig, axs = plt.subplots(1,2, figsize=(7,3.5), sharex=True, sharey=True)

for j, (key, da) in enumerate(data.items()):
    ax = axs[j]
    im = da.plot.pcolormesh(**plot_opts,ax=ax)
    ax.set_title(dataset_legends[key])
    ax.set_aspect('equal', adjustable='box')

plt.tight_layout()
fig.colorbar(im, ax=axs[:], location='right', shrink=0.8)

plt.savefig("fig6_slices.svg", transparent=True, bbox_inches="tight")
plt.savefig("fig6_slices.pdf", transparent=True, bbox_inches="tight")
plt.savefig("fig6_slices.png", transparent=True, bbox_inches="tight", dpi=300)
```
