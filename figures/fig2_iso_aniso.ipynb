# Figure 2 iso/aniso plots

```python
import pandas as pd
import matplotlib.pyplot as plt

DATASETS = {
    'CHESS 2025-09-24':'chess_20250924',
    'ALS 2025-10-15':'als_20251015',
    'CHESS 2025-11-05':'chess_20251105',
    'ALS 2025-11-25':'als_20251125',
}

data = {}

# load all the merged data
for dset_key, dset_root in DATASETS.items():
    data[dset_key] = pd.read_csv(f'source_data/fig2_iso_aniso_{dset_root}.csv').set_index('s')
```

```python
fig, axs = plt.subplots(2, 1, sharex=True, figsize=(4.25,5.5))

# add vertical line annotations
for resol in [6,3.3,2.2,1.7]:
    axs[0].axvline(x=1/resol, color='gray', linestyle='--')
    axs[1].axvline(x=1/resol, color='gray', linestyle='--')
    axs[0].text(x=1/resol + 0.01, y=0.1, s=f"{resol} Å", color='black',
            transform=axs[0].get_xaxis_transform(),
            verticalalignment='center')

# plot curves
for dset_key, df in data.items():
    df['intensity_iso'].plot(ax=axs[0])
    df['intensity_aniso'].plot(ax=axs[1],label=dset_key)

# add text labels
axs[1].legend(loc="upper left", ncols=2, bbox_to_anchor=(-.25,-.25))
axs[1].set_xlim([0,0.8])
axs[0].set_ylim([0,1.05])
axs[1].set_ylim([0,0.065])
axs[1].set_xlabel('s (Å$^{-1}$)')
axs[0].set_ylabel('Intensity\n(normalized)')
axs[1].set_ylabel('R.M.S. Intensity\n(normalized, corrected)')
plt.tight_layout()

# save as svg, pdf, and high resolution png
plt.savefig("fig2_iso_aniso.svg", transparent=True, bbox_inches="tight")
plt.savefig("fig2_iso_aniso.pdf", transparent=True, bbox_inches="tight")
plt.savefig("fig2_iso_aniso.png", transparent=True, dpi=300)
```
