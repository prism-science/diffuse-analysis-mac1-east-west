```python
import pandas as pd
import matplotlib.pyplot as plt

dataset_labels = {
    'ALS 2025-10-15 0.05s': 'als_20251015_50ms',
    'ALS 2025-10-15 0.1s': 'als_20251015_100ms',
}

data = {}

# load all the merged data
for key, label in dataset_labels.items():
    data[key] = pd.read_csv(f'source_data/fig5_iso_aniso_{label}.csv').set_index('s')
```

```python
plotopts = {
    ('ALS 2025-10-15 0.05s',0):{'label':'0.05s first half', 'color':'tab:green'},
    ('ALS 2025-10-15 0.05s',1):{'label':'0.05s second half', 'color':'lightgreen', 'linestyle':'--'},
    ('ALS 2025-10-15 0.1s',0):{'label':'0.1s first half', 'color':'rebeccapurple'},
    ('ALS 2025-10-15 0.1s',1):{'label':'0.1s second half', 'color':'plum', 'linestyle':'--'},
}

fig, axs = plt.subplots(1,2,figsize=(9,3),sharex=True)

for key, df in data.items():
    for j, col in enumerate(['intensity_iso_1','intensity_iso_2']):
        df[col].plot(ax=axs[0], **plotopts[(key,j)])
    for j, col in enumerate(['irmscorr_1','irmscorr_2']):
        df[col].plot(ax=axs[1], **plotopts[(key,j)])

for ax in axs:
    ax.set_xlim(0, 0.8)
    ax.set_xlabel("s (Å$^{-1}$)")
        
axs[0].set_ylabel("Intensity\n(normalized)")
axs[0].set_ylim(0, 1.05)
axs[0].set_title("Isotropic")

axs[1].set_ylabel("R.M.S. intensity\n(normalized, corrected)")
axs[1].set_ylim(0, 0.075)
axs[1].set_title("Anisotropic")

handles, labels = axs[1].get_legend_handles_labels()
fig.legend(handles, labels, loc="lower center", ncols=2, bbox_to_anchor=(.5,-.15))

plt.tight_layout()

# save as svg, pdf, and high resolution png
plt.savefig("export/fig5_iso_aniso.svg", transparent=True, bbox_inches="tight")
plt.savefig("export/fig5_iso_aniso.pdf", transparent=True, bbox_inches="tight")
plt.savefig("export/fig5_iso_aniso.png", transparent=True, bbox_inches="tight", dpi=300)
```
