```python
import pandas as pd

df = pd.read_csv('source_data/fig3_combo_snr_chess_20250924.csv').set_index('s')
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(1, 1, figsize=(7,3.5))

N_COLORS = {1: 'tab:blue', 2: 'tab:orange', 3: 'tab:green'}

seen = set()
for col_name, col_data in df.items():
    n = len(col_name.split('+'))
    ax.plot(col_data.index, col_data, color=N_COLORS[n], lw=1.6, alpha=0.85, 
             label=f'n = {n}' if n not in seen else None)
    seen.add(n)
    
ax.axhline(1, color='k', ls='--', lw=1, alpha=0.7, label='SNR=1')
ax.set_xlim(0, 0.8)
ax.set_ylim(0.5, 10)
ax.set_yscale('log')
ax.legend()
ax.set_xlabel("s (Å$^{-1}$)")
ax.set_ylabel('Corrected RMS intensity / RMS error')

plt.tight_layout()

plt.savefig("fig3_snr.svg", transparent=True, bbox_inches="tight")
plt.savefig("fig3_snr.pdf", transparent=True, bbox_inches="tight")
plt.savefig("fig3_snr.png", transparent=True, dpi=300)
```
