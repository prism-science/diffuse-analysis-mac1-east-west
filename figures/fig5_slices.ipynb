```python
import xarray as xr
import json

# load source data

dataset_labels = {
    'ALS 2025-10-15 0.05s': 'als_20251015_50ms',
    'ALS 2025-10-15 0.1s': 'als_20251015_100ms',
}

data = {}
for key, label in dataset_labels.items():
    fn = f'source_data/fig5_slices_{label}.json'
    with open(fn,'r') as f:
        print('loading',fn)
        data[key] = xr.Dataset.from_dict(json.load(f))
```

```python
import matplotlib.pyplot as plt

plot_opts = dict(
    x="h", 
    y="k", 
    vmin=-0.12, 
    vmax=0.12, 
    cmap='RdBu_r',
    add_colorbar=False,
    rasterized=True,
)
fig, axs = plt.subplots(2,2, figsize=(7,6), sharex=True, sharey=True)

labels = {
    ('ALS 2025-10-15 0.05s', 'Part 1'):'0.05s, first half',
    ('ALS 2025-10-15 0.05s', 'Part 2'):'0.05s, second half',
    ('ALS 2025-10-15 0.1s', 'Part 1'):'0.1s, first half',
    ('ALS 2025-10-15 0.1s', 'Part 2'):'0.1s, second half',
}

for row, (key, da) in enumerate(data.items()):
    for col, name in enumerate(['Part 1','Part 2']):
        ax = axs[row,col]
        im = da[name].plot.pcolormesh(**plot_opts,ax=ax)
        ax.set_title(labels[(key, name)])
        ax.set_aspect('equal', adjustable='box')

plt.tight_layout()
fig.colorbar(im, ax=axs[:, :], location='right', shrink=0.8)

plt.savefig("fig5_slices.svg", transparent=True, bbox_inches="tight")
plt.savefig("fig5_slices.pdf", transparent=True, bbox_inches="tight")
plt.savefig("fig5_slices.png", transparent=True, bbox_inches="tight", dpi=300)
```
