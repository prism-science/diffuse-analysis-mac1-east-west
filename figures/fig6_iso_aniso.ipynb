```python
import pandas as pd

dataset_labels = {
    'CHESS 2025-11-05': 'chess_20251105',
    'ALS 2026-04-23': 'als_20260423',
}

data = {}
for key, label in dataset_labels.items():
    fn = f'source_data/fig6_iso_aniso_{label}.csv'
    print('loading',fn)
    data[key] = pd.read_csv(fn).set_index('s')
```

```python
import matplotlib.pyplot as plt

fig, axs = plt.subplots(1,2,figsize=(9,3),sharex=True)

dataset_legends = {
    'CHESS 2025-11-05': 'CHESS 2025-11-05 (large xtals)',
    'ALS 2026-04-23': 'ALS 2026-04-23 (small xtals)',
}

for key, df in data.items():
    df['irmscorr'].plot(ax=axs[1],label=dataset_legends[key])
    df['intensity_iso'].plot(ax=axs[0],label=dataset_legends[key])

for ax in axs:
    ax.set_xlim(0, 0.8)
    ax.set_xlabel("s (Å$^{-1}$)")
        
axs[0].set_ylabel("Intensity\n(normalized)")
axs[0].set_ylim(0, 1.05)
axs[0].set_title("Isotropic")

axs[1].set_ylabel("R.M.S. intensity\n(normalized, corrected)")
axs[1].set_ylim(0, 0.07)
axs[1].set_title("Anisotropic")

handles, labels = axs[1].get_legend_handles_labels()
fig.legend(handles, labels, loc="lower center", ncols=2, bbox_to_anchor=(.5,-.1))

plt.tight_layout()

# save as svg, pdf, and high resolution png
plt.savefig("export/fig6_iso_aniso.svg", transparent=True, bbox_inches="tight")
plt.savefig("export/fig6_iso_aniso.pdf", transparent=True, bbox_inches="tight")
plt.savefig("export/fig6_iso_aniso.png", transparent=True, bbox_inches="tight", dpi=300)
```
